**Numerical Summaries**

In class, we discussed summarizing data in the column of a DataFrame. For example, consulting the data dictionary (above), we see that the `AnimalGroupParent` column tells you what kind of animal was rescued. Dog? Cat? Horse? That makes it a *qualitative variable*. A good way to summarize such data is by simply counting the number of times each value (each kind of animal) occurs.

In this case, how many cats? How many birds? How many snakes? As we saw in our previous homework, we can use `.value_counts()` -- it is a method of a Series, a single column.

````python
# focus on the AnimalGroupParent column and then count
# the number of different animal types present

lfb["AnimalGroupParent"].value_counts()
````

The lesson here is not to own a cat? How could we calibrate these figures? Do more people own cats as pets than birds or dogs, and hence the larger number of rescues? Or are cats just more complicated pets?

You see `Cat` and `cat` in this list. Presumably both refer to the same kind of animal and so here is an example of where we want to "clean" the data a bit and perhaps switch everything to upper or lower  case so that these two sets of rows get counted together. 🤔🤔🤔

**Back to counting cases**

You can also compute this kind of summary  using a DataFrame directly. That is, a DataFrame also has a method `.value_counts()`. So like `.head()` and `.tail()`, both a Series (a single column) and a DataFrame can tally up unique entries with `.value_counts()`.

For example, let's make a subset of the `lfb` focusing on the `PumpCount` and the kind of animal it is.

````python
# create a DataFrame with two columns and call it animals
animals = lfb[["AnimalGroupParent","PumpCount"]]

# have a look
animals.head()
````

Having created a two-column DataFrame `animals`, now try `.value_counts()`.

````python
animals.value_counts()
````

In general, you can think of `value_counts()` called in this way as grouping data by the unique combinations of values of the rows from a DataFrame, and then counting how many of each combination we have. We then get a `Series` that holds the counts for each unique combination.

Notice we have three cat rescues, each needing 3 trucks!

This kind of simple counting works well as a summary because we have a *qualitative variable*. It probably wouldn't be so effective with quantitative data that will presumably have a large number of unique values. More on these distinctions when we start making graphics.

As an aside, you could create both of the `.value_counts()` displays directly from `lfb` without forming subsets first. In some cases, forming a subset first is awkward and so, just as you did with subsetting, you can instead invoke `.value_counts()` on the full DataFrame, passing either the name of a column or a list of column names to summarize over.

So, for `lfb` this code provides the counts of animal types...


````python
lfb.value_counts("AnimalGroupParent")
````

... and this looks at combinations of animal type and `PumpCount`.

````python
lfb.value_counts(["AnimalGroupParent","PumpCount"])
````

We noticed that there were missing values in some of the rows of the data set `lfb`. Now, for the `IncidentNotionalCost(£)`, we computed means and sums, hoping to get a sense of the overall amount spent on animal rescues. By default, Pandas will ignore  missing values and compute averages or totals by leaving them out. When a column has a lot of missing values, these summaries might not be representative of the true costs, say. (Perhaps all the rescues from one rural neighborhood are missing, but it's in rural areas that horses and other expensive animals need the fire brigade.)

If we want to know how many missing values are in a column, we can use the method `.isna()` which returns a Boolean Series that is `True` if the value in a row is missing and `False` otherwise.

````python
lfb['IncidentNotionalCost(£)'].isna()
````

The result is a Series of booleans. It's a column where each row is `True` if the corresponding value is missing and `False` otherwise. We can use the method `.sum()` to add up the entries. In this case `True` is 1 and `False` becomes 0. So the code below tells you how many rescues are missing costs associated with them.

````python
lfb['IncidentNotionalCost(£)'].isna().sum()
````

Notice how we are just stringing the methods together. The `.isna()` returns a Series which has a summary method `.sum()`. Ha!

This says we have 106 missing values out of 14,361 total rescues. Not bad.

As with the other methods we've seen so far,  `.isna()` is also available for DataFrames as well as a Series. So instead of asking for missing values one at a time, we can apply it to the entire DataFrame.

````python
# who has something to hide?

lfb.isna()
````

So `.isna()`, applied from a DataFrame returns a DataFrame of the same size as the original, `lfb`, but it is made up of Booleans indicating which entries are missing.

The summaries we saw for columns (`.sum()` and `.mean()`, etc.) can be applied across a DataFrame too. Here we take a sum as we did before to tell us how many missing values there are per column.

````python
lfb.isna().sum()
````

You see that there are lots of missing location variables (`Easting_m` and `Northing_m` and `Latitude` and `Longitude`). Why?

We have exhibited a number of different ways of describing both qualitative data a la `.value_counts()`, as well as quantitative data through the classical statistical summaries. **Subsetting, again**

Subsetting columns of a DataFrame lets us focus attention on a few of the measurements in a table. But this is only one kind of reduction -- suppose, instead, we want to highlight only certain rows in a DataFrame.

<img src="https://github.com/cocteau/CJ2022/raw/main/images/t3.jpeg" width=300>



One way to select rows is to specify a *slice* as you did with lists. We again use square brackets but now with row selections `m:n` for example.

````python
# select row 11 (we start counting from 0) up to but not including
# row 16
lfb[10:15]
````

This is pretty limited, but is sometimes of use.

More commonly, we will  want to study one kind of animal, performing an analysis on just a subset of the rows corresponding to cats, say, or to dogs.

For subsetting rows in this way, we will create a Series of Booleans, with as many entries as a DataFrame has rows. We will then make a new DataFrame that has just the rows associated with a `True` value, and drop those with a `False`.

Let's test if each animal is a dog or not. Here are the first and last 5 animals from `lfb`.

````python
lfb["AnimalGroupParent"]
````

And here is a column of Booleans, one for each row of `lfb`. Notice that the first and third entries are `True` because the first and third entries above are dogs.

````python
lfb["AnimalGroupParent"] == "Dog"
````

The `==` is called a *logical operator* and it is making a comparison between the animal type in each row and whether or not it equals "Dog". That's cap-D, low-a, low-g exactly.

There are various logical operators

There are other logical operators like

|Operator| In words|
|----|-----|
|`!=`| not equal|
|`>` |greater than|
| `<` |less than|
| `<=` |less than or equal to|


Let's try them out. Here's the price for a rescue again, and let's test which are more than 300.

````python
lfb["IncidentNotionalCost(£)"]
````

````python
lfb["IncidentNotionalCost(£)"] > 300
````

You see that the first rescue is 510 which is larger than 300, while the next four are all 255 and so less than 300. The patterns of `True` and `False` make sense.

You can combine logical expressions with `&` for "and", as in dogs that cost more than 300 to resuce.

```
(lfb["AnimalGroupParent"] == "Dog") & (lfb["IncidentNotionalCost(£)"] > 300)
```

You can indicate Dogs or Cats using `|` for "or".

```
(lfb["AnimalGroupParent"] == "Dog") | (lfb["AnimalGroupParent"] == "Cat")
```

And finally you indicate "not" with `~`, as in animals that are not dogs.

```
~(lfb["AnimalGroupParent"] == "Dog")
```

In these expressions, like those you saw in algebra, we use parentheses to group operations. Think of something like...

```
(3+5)+(10*7+2)
```

Let's put this to work. If you want to find out how many dogs we have, we can use the function `sum()` on our column of Booleans. Again, the function `.sum()` will turn the `True`'s to a 1 and the `False`'s to 0.

Below we give the Boolean Series a name `isdog` and then use the `.sum()` method.

````python
isdog = lfb["AnimalGroupParent"] == "Dog"
isdog.sum()
````

The group-by operation extends this idea, and lets us aggregate our data based on some label or the values of one or more columns in our data. In words, the group-by operation *splits* the data according to some condition, *applies* some aggregation within those groups, and then *combines* the data back together.

In pictures, this is how group-by works.

![Split Apply Combine figure](https://github.com/altair-viz/altair-tutorial/blob/master/notebooks/split-apply-combine.png?raw=1)
[Figure source](https://jakevdp.github.io/PythonDataScienceHandbook/03.08-aggregation-and-grouping.html)

So let's look at how pandas implements this group-by operation using the method `groupby()`. First, let's split the data into groups based on `AnimalGroupParent`. We'll just use the `lfb` DataFrame, and call `groupby()` with this column name.



````python
pet_groups = lfb.groupby("AnimalGroupParent")
pet_groups
````

As we have seen, this returns a `DataFrameGroupBy` object. Stepping ahead just a little, we can examine what this structure looks like. Literally it has created groups by dividing the `lfb` DataFrame into pieces. One piece per animal type in the `AnimalGroupParent` column.

To see this we will *iterate* or *loop* over the groups using a `for` loop. Let's run it and talk about what it is doing after.

````python
# loop over the groups formed by groupby(). Each has a
# label (a kind of pet) and then a DataFrame holding all
# the rows relating to a rescue of this kind of pet.

for pet,df in pet_groups:
  print("---"*10)
  print("First five rows of group for pets -", pet)
  print(df.head(5))
````

Each animal name (Cat, Dog, Bird) is associated with a DataFrame that has just the rows from `lfb` that involve the rescue of that kind of pet. All the cat rescues in one data frame, all the dogs in another.

The display above starts with a separator "-----------". It then has a message that what follows are the first five rows of the DataFrame associated with a particular animal. It then prints out the first five rows (assuming there are five). The printout of the rows is long and not what we're used to seeing. But scroll through the output and start with a "-----------" and read it over. Make sure you understand that we're looking at pieces of our `lfb` DataFrame broken down by animal kind.

We can grab one of these group's DataFrame using `get_group()`.

````python
pet_groups.get_group("Cat")
````

To be clear, this is all the cats from `lfb`. That should be the same as if we manually extracted just the cats from `lfb` using the subsetting we learned.

Which compares to what we would do if we manually extracted the column of incident numbers from the DataFrame of just cats.

Finally, to each group we can apply one of our summary statistics like `sum()` and `std()` and `count()`.

Here is one summary, the number of rows or the *count* of items in our aggregate, this time just applied first to the group corresponding to cats...

````python
pet_groups.get_group("Cat")["IncidentNumber"].count()
````

... and then removing `.get_group()` and applying it to all the groups.

````python
pet_groups["IncidentNumber"].count()
````

We can do these calls in one line. That is, instead of forming `pet_groups`, we can do all this in one command.

````python
lfb.groupby("AnimalGroupParent")["IncidentNumber"].count()
````

This is incredibly powerful for our analysis!

Let's push this a little farther. First, compute the number of rescues per year. We group-by `CalYear` and then `count()` the number of events in each year.

````python
lfb.groupby("CalYear")["IncidentNumber"].count()
````

The result here is a Series. The title `CalYear` is playing the role of an index labeling each entry in the Series. We can keep the `CalYear` as a column in a DataFrame by using the argument `as_index=False`.

````python
lfb.groupby("CalYear",as_index=False)["IncidentNumber"].count()
````

````python
lfb.groupby("CalYear",as_index=False)[["IncidentNotionalCost(£)","PumpHoursTotal"]].sum()
````

Finally, we can also group-by the unique combinations of values from two or more columns. This is like what we did with `.value_counts()`.

````python
lfb.groupby(["AnimalGroupParent","CalYear"],as_index=False)[["IncidentNotionalCost(£)","PumpHoursTotal"]].sum()
````

So far, we have applied one summary to all the columns we specify. Instead, we can compute any kind of summary statistic for each column. Here we use the method `.agg()` to be very explicit about what we are wanting to compute. This method takes a dictionary with column names as keys and then summary statistics as the associated values.

The result below is a new DataFrame called `costs` that has columns of animal type, total cost spent rescuing each type of animal, and the number of rescues performed. In `.agg()` we call `sum()` for `IncidentNotionalCost(£)` and `count()` for `IncidentNumber`.

````python
# use .agg() to specify how to summarize each column
costs = lfb.groupby("AnimalGroupParent",as_index=False).agg({"IncidentNotionalCost(£)":"sum",
                                                             "IncidentNumber":"count"})
costs
````

We are irresistibly led to compute the average price for each animal group. Aren't we? To do this we create a new column called `AvgPrice` as the ratio of the total cost and the total number of incidents.

Have a look!

````python
costs = lfb.groupby("AnimalGroupParent", as_index=False).agg({
    "IncidentNotionalCost(£)": "sum",
    "IncidentNumber": "count"
})

costs["AvgPrice"] = (
    costs["IncidentNotionalCost(£)"] / costs["IncidentNumber"]
)

costs
````

````python
costs.sort_values("AvgPrice",ascending=False)
````
